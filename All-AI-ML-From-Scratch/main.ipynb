```python
from models.linear_regression import LinearRegression
```

```python
model = LinearRegression(learning_rate=0.01, lambda_=0.002, iterations=500)
```

```python
import numpy as np

# Feature: house size (sqft)
X = np.array([
    [850],
    [900],
    [1200],
    [1500],
    [1430],
    [1800],
    [1350],
    [1080],
    [2000]
])

# Target: house price
y = np.array([
    150000,
    160000,
    200000,
    250000,
    245000,
    300000,
    220000,
    300000,
    300000,
    340000
])
```

```python
from sklearn.preprocessing import StandardScaler

x_scaler = StandardScaler()
y_scaler = StandardScaler()

X_scaled = x_scaler.fit_transform(X)

y_scaled = y_scaler.fit_transform(
    y.reshape(-1, 1)
).flatten()

model.fit(X_scaled, y_scaled)
```

```python
y_scaled, X_scaled
```

```python
import matplotlib.pyplot as plt

plt.plot(model.cost_history)
plt.xlabel("Iterations")
plt.ylabel("Cost")
plt.title("Training Convergence")
plt.show()
```

```python
plt.scatter(X_scaled, y_scaled, color="blue")

plt.plot(X_scaled, model.predict(X_scaled), color="red")

plt.xlabel("House Size")
plt.ylabel("Price")
plt.title("Linear Regression Fit")

plt.show()
```

```python
new_house = np.array([[1700]])

prediction = model.predict(new_house)

print(prediction)
```
