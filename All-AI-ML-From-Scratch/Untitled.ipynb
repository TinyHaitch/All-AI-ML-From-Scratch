```python
from models.linear_regression import LinearRegression
```

```python
model = LinearRegression(learning_rate=0.000001, lambda_=0.2, iterations=1000)
```

```python
import numpy as np

# Feature: house size (sqft)
X = np.array([
    [850],
    [900],
    [1200],
    [1500],
    [1800],
    [2000]
])

# Target: house price
y = np.array([
    150000,
    160000,
    200000,
    240000,
    300000,
    340000
])
```

```python
model.fit(X, y)
```

```python
import matplotlib.pyplot as plt

plt.plot(model.cost_history)
plt.xlabel("Iterations")
plt.ylabel("Cost")
plt.title("Training Convergence")
plt.show()
```
